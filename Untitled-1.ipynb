```python
log_entry = "   [ERROR] status_code=404 | path=/api/v1/users | client=192.168.1.1 | duration=120ms   "

# 1. Очищаем пробелы и берем часть после [ERROR]
clean_data = log_entry.strip().partition("[ERROR]")[2].strip()

# Разбиваем полученную строку по символу "|"
# clean_data станет: "status_code=404 | path=/api/v1/users | client=192.168.1.1 | duration=120ms"
parts = clean_data.split(" | ")

# Извлекаем и преобразуем статус в int (часть после "=")
status_code = int(parts[0].split("=")[1])
print(status_code)

# 2. Достаем путь (/api/v1/users) и вытаскиваем версию (второй элемент при split по "/")
path = parts[1].split("=")[1]
api_version = path.split("/")[2]

# 3. Достаем время (120ms), отрезаем "ms" и переводим в int
duration_str = parts[3].split("=")[1]
duration_ms = int(duration_str.rstrip("ms"))

# 4. Собираем итоговый словарь
parsed_log = {
    "status": status_code,
    "api_version": api_version,
    "duration_ms": duration_ms
}

print(parsed_log)
# Выведет: {'status': 404, 'api_version': 'v1', 'duration_ms': 120}
```

```python
l = [1,1,1,1,1,1,1,1,5,5,5,5,5,3,3,3,2,2,1]

my_dict = {}
pairs = []

for i in l:
    if i in my_dict:
        my_dict[i]+=1
    else:
        my_dict[i] = my_dict.get(i,1)
print(my_dict)

for key,item in my_dict.items():
    pairs.append((item,key))

pairs.sort()
print(pairs)
```
