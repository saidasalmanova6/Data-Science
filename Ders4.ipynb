# Functions

A **function** is a named block of code that does one job and can be used again and again.

Why we use them:

- **Write once, use many times** – no copy-pasted code.
- **Shorter and clearer** – the name says what the block does.
- **Easier to fix** – a bug is corrected in one place only.
- **Easier to test** – each piece can be checked on its own.

### Kinds of functions in Python

| Kind | What it is | Example |
|---|---|---|
| Built-in | already inside Python | `print()`, `len()`, `sum()`, `type()` |
| User-defined | written by us with `def` | `def greet(): ...` |
| Anonymous (lambda) | a one-line function with no name | `lambda x: x ** 2` |
| Higher-order | takes or returns another function | `map()`, `filter()`, `sorted()` |
| Generator | produces values one by one with `yield` | `def fib(): ... yield a` |
| Recursive | calls itself | `def factorial(n): ... factorial(n-1)` |

We will go through all of them in this lesson.

## 1. Defining and calling a function

```python
def function_name(parameter1, parameter2):
    """docstring – what the function does"""
    # function body
    return result
```

| Term | Meaning |
|---|---|
| `def` | the keyword that starts a definition |
| **parameter** | the name in the definition (`x`) |
| **argument** | the real value given at the call (`5`) |
| body | the indented block – the code that runs |
| `return` | sends a value back and ends the function |

Defining a function does **not** run it. It runs only when it is called: `function_name(...)`.

### 1.1 `return` is not `print`

This is the most important distinction of the whole lesson.

| | `print()` | `return` |
|---|---|---|
| What it does | shows text on the screen | gives a **value back** to the caller |
| Can the result be stored? | no (`x` becomes `None`) | yes (`x = f(...)`) |
| Can it be used in a calculation? | no | yes |
| Does it end the function? | no | **yes** |

A function without `return` gives back `None`.

### 1.3 Docstrings and `help()`

A **docstring** is a text in triple quotes written right after the `def` line. It explains what
the function does, what it takes and what it returns. `help(function)` shows it, and it is also
stored in `function.__doc__`.

````python
def average(numbers):
    """
    Return the average of a list of numbers.

    Parameters
    ----------
    numbers : list
        A list of int or float values.

    Returns
    -------
    float
        The arithmetic mean of the values.
    """
    return sum(numbers) / len(numbers)

print(average([2, 4, 6]))
help(average)
print(average.__doc__[:40])
````

### 1.4 Typical beginner errors

| Error | Cause |
|---|---|
| `NameError: name 'f' is not defined` | the function is called **before** it is defined |
| `TypeError: missing 1 required positional argument` | too few arguments |
| `TypeError: takes 2 positional arguments but 3 were given` | too many arguments |
| `IndentationError` | the body is not indented |
| the result is `None` | `print` was used inside instead of `return` |

Also: never give a function (or a variable) the name of a built-in – `sum`, `list`, `len`,
`str`, `max`. After `sum = 0` the built-in `sum()` stops working in that session.

````python
# a function must be defined before it is called
# hello()        # NameError: name 'hello' is not defined

def hello():
    pass         # pass = an empty body, a placeholder for later

hello()
print(hello())   # None - the body does nothing

total = sum([1, 2, 3])    # the built-in still works
print(total)
````

## 2. Parameters and arguments

Python has several ways of giving values to a function. This section goes through all of them
and the rules that connect them.

### 2.1 Positional arguments

!!! The values are matched to the parameters **by their position**. The order matters and the count
must be exact.

````python
def bolek(a,b):
    return a/b

print(bolek(10,2))
print(bolek(2,10))

print(bolek(b =2,a =10))
````

### 2.2 Default values

A parameter can be given a default value in the definition. Then the argument becomes optional.

**Rule:** parameters with a default must come **after** the ones without a default, otherwise
Python raises `SyntaxError: parameter without a default follows parameter with a default`.

````python
def hesablama(x,kvadrat=True):
    if kvadrat:
        return x**2

    return x*2
hesablama(10,kvadrat="0")
````

````python
# This is invalid - written as a string so that the notebook still runs:
bad = """
def myfunc(p=2, x):      # SyntaxError: parameter without a default follows one with a default
    return x ** p
"""
print(bad)

def good(x, p=2):        # correct order
    return x ** p

print(good(3))
````

### 2.3 Keyword arguments

If the names are written at the call, the order does not matter any more. This makes calls with
many optional parameters much easier to read.

Rule: positional arguments must come **before** keyword arguments in a call.

````python
def power(x, p=2, debug=False):
    if debug:
        print(f"Calculating {x} ** {p}")
    return x ** p

print(power(p=3, debug=True, x=7))    # any order
print(power(7, debug=True))           # mixed: positional first
# print(power(x=7, 3))                # SyntaxError: positional argument follows keyword argument
````

### 2.4 The trap: a mutable default value

A default value is created **once**, when the function is defined – not at every call. If that
default is a list (or a dict or a set), every call shares the **same** list, so the values pile
up. This is one of the best known Python traps.

The fix is always the same: use `None` as the default and create the list inside.

### 2.7 The order of the parameters

In a definition the order is fixed:

| Order | Parameter type | Example |
|---|---|---|
| 1 | positional | `a, b` |
| 2 | with a default | `p=2` |
| 3 | `*args` | `*args` |
| 4 | keyword-only (after `*args`) | `sep=", "` |
| 5 | `**kwargs` | `**kwargs` |

Anything placed in the wrong order is a `SyntaxError`.

````python
def full(a, b=2, *args, sep="-", **kwargs):
    print("a      :", a)
    print("b      :", b)
    print("args   :", args)
    print("sep    :", sep)
    print("kwargs :", kwargs)

full(1)
print("---")
full(1, 5, 10, 20, sep="+", x=100, y=200)

# invalid order (kept as a string so the notebook still runs):
print("""
def bad(**kwargs, *args, default='Yes'):   # SyntaxError
    ...
""")
````

A parameter written after `*args` (or after a lone `*`) can **only** be given by name – this is
called a keyword-only parameter. It is useful for options that should always be explicit.

````python
def join_words(*words, sep=" "):
    return sep.join(words)

print(join_words("IT", "Brains"))
print(join_words("IT", "Brains", sep="-"))
# print(join_words("Turing", "Academy", "-"))   # "-" would just become another word
````

### 2.8 `*` and `**` at the call – unpacking

The same two symbols do the **opposite** job when they are used in a call:

| Where | `*` | `**` |
|---|---|---|
| In the definition | packs the extra arguments **into** a tuple | packs them **into** a dict |
| In the call | unpacks a list/tuple **into** separate arguments | unpacks a dict **into** keyword arguments |

````python
def my_func(arg1, arg2, arg3):
    print("arg1:", arg1)
    print("arg2:", arg2)
    print("arg3:", arg3)

args = ("Itbrains", "for", "all")
my_func(*args)                # same as my_func("Turing", "for", "all")

kwargs = {"arg2": "Itbrains", "arg1": "for", "arg3": "all"}
my_func(**kwargs)             # the keys must match the parameter names
````

### 2.9 How are the arguments passed?

Python always passes a **reference to the object**. What differs is whether the object can be
changed:

| Object type | Examples | What happens inside the function |
|---|---|---|
| **Immutable** | `int`, `float`, `str`, `tuple`, `bool` | it cannot be changed; an assignment creates a new local object, the caller sees nothing |
| **Mutable** | `list`, `dict`, `set` | methods like `append()` change the **original** object, and the caller sees it |

````python
def modify_immutable(x):
    x += 10                     # a new local value
    print("Inside function:", x)

num = 5
modify_immutable(num)
print("Outside function:", num)     # 5 - unchanged
````

````python
def modify_mutable(lst):
    lst.append(4)               # the original list is changed
    print("Inside function:", lst)

my_list = [1, 2, 3]
modify_mutable(my_list)
print("Outside function:", my_list)   # [1, 2, 3, 4] - changed!
````

````python
# two ways to avoid a surprise
def safe_append(lst, value):
    new_list = lst.copy()       # work on a copy (Lesson 1)
    new_list.append(value)
    return new_list

def rebind(lst):
    lst = [99]                  # rebinding only changes the local name
    print("inside:", lst)

original = [1, 2, 3]
print(safe_append(original, 4), original)
rebind(original)
print("outside:", original)
````

## 3. Functions are objects

In Python a function is a value like any other: it can be stored in a variable, put into a list
or a dictionary, passed to another function and returned from one. A function that takes or
returns a function is called a **higher-order function**.

Note the difference: `greet` is the function itself, `greet()` **calls** it.

### 3.2 A function that returns a function

The inner function remembers the values of the outer one – this is called a **closure**. It is

## 6. Scope: where a variable lives

A variable created **inside** a function exists only inside it. Python looks for a name in this
order (the LEGB rule):

| Order | Scope | Meaning |
|---|---|---|
| L | Local | inside the current function |
| E | Enclosing | inside the outer function (for nested functions) |
| G | Global | at the top level of the notebook / file |
| B | Built-in | Python's own names (`print`, `len`, `sum`) |

````python
message = "I am global"

def greeting():
    message = "I am local"        # a NEW local variable, the global one is untouched
    print("Inside :", message)

greeting()
print("Outside:", message)
````

````python
def greeting2():
    local_only = "Hello"
    print("Inside :", local_only)

greeting2()
# print(local_only)      # NameError: name 'local_only' is not defined
````

A global variable can be **read** inside a function without anything special. Problems start
only when we try to change it.

````python
tax = 0.18                      # global

def price_with_tax(price):
    return price * (1 + tax)    # reading the global is fine

print(price_with_tax(100))
````

### 6.1 `global`

`global` says: "this name refers to the variable outside, do not create a local one".

It works, but it is best avoided: a function that changes global state is hard to test and hard
to follow. Taking the value as a parameter and giving the new one back with `return` is almost
always better.

````python
counter = 0

def increase_bad():
    global counter          # works, but the function now has a side effect
    counter += 1

increase_bad()
increase_bad()
print(counter)

# the preferred way: in through the parameter, out through return
def increase_good(value):
    return value + 1

counter2 = 0
counter2 = increase_good(counter2)
counter2 = increase_good(counter2)
print(counter2)
````

### 6.2 `nonlocal`

`nonlocal` is used in a **nested** function: it points to the variable of the outer function –
neither local nor global.

````python
def outer():
    message = "local"

    def inner():
        nonlocal message         # the outer one is changed
        message = "nonlocal"
        print("inner :", message)

    def inner2():
        print("inner2:", message)   # only reading -> nonlocal is not needed

    inner()
    inner2()
    print("outer :", message)

outer()
````

## 7. Decorators

A **decorator** is a function that takes another function, wraps it in some extra behaviour and
returns the new function. It adds things like logging, timing or checking, **without touching
the original code**.

`@my_decorator` written above a `def` is only a short form of
`my_function = my_decorator(my_function)`.

````python
def my_decorator(function):
    def wrapper(*args, **kwargs):          # *args/**kwargs -> it fits any function
        print("Function is starting")
        result = function(*args, **kwargs)  # the original function is called
        print("Function is finished")
        return result                       # its result is passed on
    return wrapper

def greet(name):
    print(f"Hello, {name}!")
    return name.upper()

greet = my_decorator(greet)      # wrapping by hand
print(greet("Raul"))
````

````python
# exactly the same thing with the @ syntax
def my_decorator(function):
    def wrapper(*args, **kwargs):
        print("Function is starting")
        result = function(*args, **kwargs)
        print("Function is finished")
        return result
    return wrapper

@my_decorator
def add(a, b):
    return a + b

print(add(2, 3))
````

Two details that are easy to miss:

- The wrapper must **return** the result of the original function, otherwise every decorated
  function starts returning `None`.
- After decorating, `function.__name__` becomes `"wrapper"`. `functools.wraps` keeps the
  original name and docstring.

````python
import functools
import time

def timer(function):
    @functools.wraps(function)          # keeps the name and the docstring
    def wrapper(*args, **kwargs):
        start = time.time()
        result = function(*args, **kwargs)
        print(f"{function.__name__} took {time.time() - start:.4f} seconds")
        return result
    return wrapper

@timer
def slow_sum(n):
    """Add up the numbers from 0 to n."""
    return sum(range(n))

print(slow_sum(1_000_000))
print(slow_sum.__name__, "|", slow_sum.__doc__)
````

## 8. Generator functions (`yield`)

A generator function uses **`yield`** instead of `return`. It produces the values one at a time,
only when they are asked for, and remembers where it stopped.

| | `return` | `yield` |
|---|---|---|
| What happens | the function ends | it pauses and waits |
| How many times | once | as many times as needed |
| What the call gives back | the value | a **generator object** |
| Memory | the whole result is built | one value at a time |

This is the function version of the generator expressions of Lesson 2.

````python
def count_to_three():
    print("   (start)")
    yield 1
    print("   (after the first yield)")
    yield 2
    yield 3

gen = count_to_three()
print(gen)                   # nothing has run yet!

print(next(gen))             # runs up to the first yield
print(next(gen))
print(next(gen))
# print(next(gen))           # StopIteration: there is nothing left
````

````python
# a for loop calls next() for us and stops at StopIteration
def squares(n):
    for i in range(n):
        yield i ** 2

for value in squares(5):
    print(value, end=" ")
print()

print(list(squares(5)))      # it can be turned into a list
print(sum(squares(5)))
````

### 8.1 An endless sequence

A generator may run forever, because the values are produced only on demand. A normal function
could never return an infinite list.

````python
def fibonacci():
    """Generate Fibonacci numbers endlessly."""
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

fib_gen = fibonacci()

for i in range(10):
    print(next(fib_gen), end=" ")
print()
````

### 8.2 Memory and single use

The generator keeps only the current value, so its memory usage does not grow with the size of
the data – but it can be walked through only **once**.

````python
import sys

big_list = [x ** 2 for x in range(100_000)]
big_gen = (x ** 2 for x in range(100_000))

print("list     :", sys.getsizeof(big_list), "bytes")
print("generator:", sys.getsizeof(big_gen), "bytes")

gen = squares(3)
print(list(gen))
print(list(gen))          # empty: it is already used up
````

## 9. Recursive functions

A **recursive** function calls itself. Every recursion needs two parts:

1. the **base case** – when it stops (without it: `RecursionError`),
2. the **recursive case** – the same problem, one step smaller.

````python
def factorial(n):
    if n <= 1:            # base case
        return 1
    return n * factorial(n - 1)      # recursive case

print(factorial(5))       # 5 * 4 * 3 * 2 * 1
print(factorial(0))
````

````python
def countdown(n):
    if n == 0:
        print("Go!")
        return
    print(n)
    countdown(n - 1)

countdown(3)
````

### 9.1 Recursion or a loop?

The same job can almost always be written with a loop, and the loop version is usually faster.
The recursive Fibonacci calculates the same values again and again, so it becomes extremely slow
as `n` grows.

````python
def fibonacci_recursive(n):
    if n <= 0:
        return 0
    if n == 1:
        return 1
    return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)

def fibonacci_iterative(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    prev, curr = 0, 1
    for _ in range(2, n + 1):
        prev, curr = curr, prev + curr
    return curr

print(fibonacci_recursive(10), fibonacci_iterative(10))
````

````python
# compare the speed (this cell takes a few seconds)
import time

start = time.time()
fibonacci_recursive(28)
print(f"recursive: {time.time() - start:.4f} s")

start = time.time()
fibonacci_iterative(28)
print(f"iterative: {time.time() - start:.6f} s")
````

Two more things worth knowing:

- Python stops at about 1000 nested calls (`RecursionError: maximum recursion depth exceeded`).
- `functools.lru_cache` remembers the results, which makes the recursive version fast again.

````python
import sys
from functools import lru_cache

print("recursion limit:", sys.getrecursionlimit())

@lru_cache(maxsize=None)
def fib_cached(n):
    if n < 2:
        return n
    return fib_cached(n - 1) + fib_cached(n - 2)

print(fib_cached(100))       # instant, even for n = 100
````

# Summary of Lesson 3 – Functions

| Tool | What it is for |
|---|---|
| `def name(params):` | define a function |
| `return` | give a value back **and** end the function |
| docstring (triple quotes) | document it; shown by `help()` |
| `p=2` | default value – optional argument |
| `f(x=7)` | keyword argument – order does not matter |
| `*args` | any number of positional arguments → tuple |
| `**kwargs` | any number of keyword arguments → dict |
| `f(*list)` / `f(**dict)` | unpacking at the call |
| `lambda x: ...` | small anonymous function |
| `map` / `filter` / `reduce` | transform / keep / combine |
| `sorted(key=...)` | sort by a computed value |
| `global` / `nonlocal` | write to an outer variable (use sparingly) |
| `@decorator` | wrap a function in extra behaviour |
| `yield` | produce the values one by one (generator) |
| recursion | a function that calls itself (needs a base case) |

Traps to remember:

- `print` inside the function ≠ `return`; without `return` the result is `None`.
- A parameter without a default cannot follow one with a default.
- `def f(items=[])` keeps the **same** list between calls → use `None`.
- A mutable argument (list, dict) can be changed inside the function; an immutable one cannot.
- `map` / `filter` give an iterator – wrap them in `list()`.
- A generator can be walked through only once.
- Recursion without a base case → `RecursionError`.
